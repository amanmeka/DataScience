Explain what happens when we do matrix multiplication.

Yes, I know I basically already asked you to do this in the notes. Don't copy and paste your anwser from before; honestly do the exercise again.

Matrix multiplication combines two matrices, A and B, into a new matrix C. It requires A's columns to match B's rows (the last dimension of A and first dimension of B). Each element in C is the dot product of a row from A and a column from B.
